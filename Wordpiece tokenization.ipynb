```python
from collections import Counter

with open("raw.txt", "r") as file:
    corpus = [line.strip() for line in file if line.strip()]
vocab = {
    "[PAD]": 0,
    "[UNK]": 1,
    "[CLS]": 2,
    "[SEP]": 3,
    "high": 4,
    "er": 5,
    "est": 6
}
def wordpiece_tokenize(word):
    tokens = []
    start = 0

    while start < len(word):

        found = None
        for end in range(len(word), start, -1):

            piece = word[start:end]

            if piece in vocab:
                found = piece
                next_start = end
                break

        if found is None:
            return ["[UNK]"]

        tokens.append(found)
        start = next_start

    return tokens
def tokens_to_ids(tokens):
    return [
        vocab.get(token, vocab["[UNK]"])
        for token in tokens
    ]
for word in corpus:

    tokens = wordpiece_tokenize(word)
    ids = tokens_to_ids(tokens)

    print("\nWord:", word)
    print("Tokens:", tokens)
    print("IDs:", ids)
```
